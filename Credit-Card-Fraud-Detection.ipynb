```python
import pandas as pd
import numpy as np
import math
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from imblearn.over_sampling import SMOTE
```

```python
# 1. Load Data
df = pd.read_csv('creditcard.csv')
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
# 2. Balance the Dataset
smote = SMOTE(random_state=42)
X_balanced, y_balanced = smote.fit_resample(X, y)
balanced_df = pd.DataFrame(X_balanced)
balanced_df['Target'] = np.array(y_balanced)
```

```python
# 3. Calculate Sample Sizes
Z = 1.96 
p = 0.5
E = 0.05

n_simple = math.ceil(( (Z**2) * p * (1-p) ) / (E**2))

S = 2
n_stratified = math.ceil(( (Z**2) * p * (1-p) ) / ((E/S)**2))

C = 5
n_cluster = math.ceil(( (Z**2) * p * (1-p) ) / ((E/C)**2))
```

```python
# 4. 5 Samples
samples = {}

# Sample 1: Simple Random Sampling
samples['Sampling1'] = balanced_df.sample(n=n_simple, random_state=1)

for sample_name, sample_data in samples.items():
    print(f"{sample_name}")
    print(sample_data.head())
```

```python
# Sample 2: Systematic Sampling
step = len(balanced_df) // n_simple
samples['Sampling2'] = balanced_df.iloc[::step].head(n_simple)
for sample_name, sample_data in samples.items():
    print(f"{sample_name}")
    print(sample_data.head())
```

```python
# Sample 3: Stratified Sampling 
samples['Sampling3'] = balanced_df.groupby('Target').sample(n=n_stratified // 2, random_state=42)
for sample_name, sample_data in samples.items():
    print(f"{sample_name}")
    print(sample_data.head())
```

```python
# Sample 4: Cluster Sampling
clusters = np.array_split(balanced_df, len(balanced_df) // C)
selected_clusters = np.random.choice(len(clusters), size=n_cluster // C, replace=False)
samples['Sampling4'] = pd.concat([clusters[i] for i in selected_clusters])
for sample_name, sample_data in samples.items():
    print(f"{sample_name}")
    print(sample_data.head())
    
```

```python
# Sample 5: Simple Random Sampling (Different Random State)
samples['Sampling5'] = balanced_df.sample(n=n_simple, random_state=42)
for sample_name, sample_data in samples.items():
    print(f"{sample_name}")
    print(sample_data.head())
```

```python
# 5. Define 5 ML Models
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
models = {
    'M1': LogisticRegression(max_iter=1000),
    'M2': KNeighborsClassifier(),
    'M3': DecisionTreeClassifier(),
    'M4': SVC(),
    'M5': RandomForestClassifier()
}
```

```python
# 6. Train and Evaluate
results = pd.DataFrame(index=models.keys(), columns=samples.keys())
scaler = StandardScaler() # Initialize the scaler

for sample_name, sample_data in samples.items():
    X_sample = sample_data.iloc[:, :-1]
    y_sample = sample_data.iloc[:, -1]
    
    # Train-test split for evaluation
    X_train, X_test, y_train, y_test = train_test_split(X_sample, y_sample, test_size=0.2, random_state=42)
    
    # Scale the features
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    for model_name, model in models.items():
        # Train on scaled data
        model.fit(X_train_scaled, y_train)
        predictions = model.predict(X_test_scaled)
        accuracy = accuracy_score(y_test, predictions) * 100
        results.loc[model_name, sample_name] = round(accuracy, 2)

# Display final table
print(results)
```
